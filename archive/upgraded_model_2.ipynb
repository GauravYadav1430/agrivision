```python
import tensorflow as tf
from tensorflow.keras import models, layers, regularizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt
import numpy as np
import os
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# Create the output folder for your PPT graphs
os.makedirs("output", exist_ok=True)

IMAGE_SIZE = 256
BATCH_SIZE = 32
CHANNELS = 3
EPOCHS = 50 

# 1. Load Data
dataset = tf.keras.preprocessing.image_dataset_from_directory(
    "PlantVillage",
    shuffle=True,
    image_size=(IMAGE_SIZE, IMAGE_SIZE),
    batch_size=BATCH_SIZE
)
class_names = dataset.class_names

# 2. Split Data Function
def get_dataset_partitions_tf(ds, train_split=0.8, val_split=0.1, test_split=0.1, shuffle=True, shuffle_size=10000):
    ds_size = len(ds)
    if shuffle:
        ds = ds.shuffle(shuffle_size, seed=12)
    train_size = int(train_split * ds_size)
    val_size = int(val_split * ds_size)
    train_ds = ds.take(train_size)
    val_ds = ds.skip(train_size).take(val_size)
    test_ds = ds.skip(train_size).skip(val_size)
    return train_ds, val_ds, test_ds

train_ds, val_ds, test_ds = get_dataset_partitions_tf(dataset)

# Optimize for speed
train_ds = train_ds.cache().shuffle(1000).prefetch(buffer_size=tf.data.AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)
test_ds = test_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)

# 3. Preprocessing Layers
resize_and_rescale = tf.keras.Sequential([
    layers.Resizing(IMAGE_SIZE, IMAGE_SIZE),
    layers.Rescaling(1.0/255)
])

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.2)
])

# 4. THE HIGH-PERFORMANCE ARCHITECTURE
input_shape = (BATCH_SIZE, IMAGE_SIZE, IMAGE_SIZE, CHANNELS)
n_classes = 3

model = models.Sequential([
    resize_and_rescale,
    data_augmentation,
    
    layers.Conv2D(32, kernel_size=(3,3), activation='relu', input_shape=input_shape),
    layers.MaxPooling2D((2, 2)),
    
    layers.Conv2D(64,  kernel_size=(3,3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    layers.Conv2D(64,  kernel_size=(3,3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    layers.Flatten(),
    
    # --- The Real Upgrades ---
    # L2 Regularizer acts as a smooth penalty on the weights to prevent overfitting
    layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
    
    # Very light dropout (10%) just to close the gap between train and val accuracy
    layers.Dropout(0.1),
    
    layers.Dense(n_classes, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False),
    metrics=['accuracy']
)

model.summary()

# 5. The Advanced Callbacks
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=8, 
    restore_best_weights=True
)

# This is the secret weapon! It lowers the learning rate when accuracy gets stuck.
lr_scheduler = ReduceLROnPlateau(
    monitor='val_loss', 
    factor=0.5,     # Cut learning rate in half
    patience=3,     # If it doesn't improve for 3 epochs
    min_lr=1e-6,
    verbose=1
)

# 6. Train the Model
print("Training the High-Performance AI Model...")
history = model.fit(
    train_ds,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_data=val_ds,
    verbose=1,
    callbacks=[early_stopping, lr_scheduler] # <--- Added the new scheduler here
)

# Save the brain
model.save("upgraded_model.h5")
print("Model saved successfully as upgraded_model.h5")

# 7. Generate & Save PPT Graphs
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(len(acc)) 

plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy', color='green', linewidth=2)
plt.plot(epochs_range, val_acc, label='Validation Accuracy', color='blue', linewidth=2)
plt.legend(loc='lower right')
plt.title('Training and Validation Accuracy')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss', color='red', linewidth=2)
plt.plot(epochs_range, val_loss, label='Validation Loss', color='orange', linewidth=2)
plt.legend(loc='upper right')
plt.title('Training and Validation Loss')
plt.grid(True)
plt.savefig("output/upgraded_learning_curves.png", dpi=300, bbox_inches='tight')
plt.show()

# Confusion Matrix
print("Generating Confusion Matrix...")
y_pred = []
y_true = []
for images_batch, labels_batch in test_ds:
    y_true.extend(labels_batch.numpy())
    batch_predictions = model.predict(images_batch, verbose=0)
    y_pred.extend(np.argmax(batch_predictions, axis=1))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title('Potato Disease Confusion Matrix')
plt.ylabel('Actual Leaf Disease')
plt.xlabel('AI Predicted Disease')
plt.savefig("output/upgraded_confusion_matrix.png", dpi=300, bbox_inches='tight')
plt.show()

# Classification Report
report = classification_report(y_true, y_pred, target_names=class_names)
print(report)
with open("output/upgraded_metrics.txt", "w") as f:
    f.write(report)
print("All PPT assets saved to the 'output' folder!")
```
